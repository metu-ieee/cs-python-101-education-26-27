#### String Metotları

Stringleri düzenlemek için elimizde oldukça çok metot var. Hadi tek tek keşfedelim! Unutmayalım ki burada yazdığımdan çok daha fazla var, ancak onlar günlük hayatta karşılaşmamızın çok düşük olduğu şeyler, dolayısıyla burada yer vermedim. Ancak [buraya tıklayıp](https://www.w3schools.com/python/python_ref_string.asp) geri kalanını görebilirsiniz. Bunları ezberlemek zorunda değilsiniz, unuttuğunuz zaman bir kez bile internette aratmak yeterli gelmekte.

**Büyük Harf - Küçük Harf değiştirenler**
* `string.capitalize()`: String'in **sadece** ilk harfini büyük yapar.
* `string.casefold()`: Tüm harfleri küçük yapar.
* `string.lower()`: Bu da tüm harfleri küçük yapar, ancak daha güçsüzdür, bunu aşağıda anlatacağım.
* `string.swapcase()`: Büyük olanları küçük, küçük olanları büyük yapar.
* `string.title()`: Her kelimenin ilk harfini büyük yapar.
* `string.upper()`: Tüm harfleri büyük harf yapar.

**Sayısal işlemler**
* `string.count("str")`: String içerisinde verilen string kaç tane onu bulur.
* `string.find("str")`: String içerisinde aranan string'in ilkinin index'ini döndürür. Bulamazsa -1 döndürür.
* `string.index("str")`: String içerisinde aranan string'in ilkinin index'ini döndürür. Bulamazsa sistem hata verir.
* `string.rfind("str")`: String içerisinde aranan string'in **sonunun** index'ini döndürür. Bulamazsa -1 döndürür.
* `string.rindex("str")`: String içerisinde aranan string'in **sonunun** index'ini döndürür. Bulamazsa sistem hata verir.

**Önemli iki sorgu**
* `string.startswith("str")`: String verilen string ile başlıyorsa *True* değilse *False* döndürür.
* `string.endswith("str")`: String verilen string ile bitiyorsa *True* değilse *False* döndürür.

**String üretme ve bölme**
* `string.join(iterable)`: Verilen liste (ya da tuple, set vs.) item'lerinin arasında string'i koyup yeni bir string türetir.
* `string.replace("str1", "str2")`: String içindeki verilen string 1'i bulur ve hepsini string 2 ile değiştirir.
* `string.split("str")`: String'i verilen string'i gördüğü her yerden böler ve bunlar ile bir liste oluşturur. Parantez boş bırakılırsa " " (boşluk) gördüğü yerlerden böler.
* `string.splitlines()`: String'i satırlara ayırır. Burada sadece `"\n"` ifadesini gördüğü yerlerden ayırır. Yani `string.splitlines() = string.split("\n")`.

***ÖNEMLİ NOT: Bu metotlar string'i direkt değiştirmez, bir string döndürür. Dolayısıyla bir değişkene eşitlemeniz gerekmektedir.***

```python
# Hadi önemli olanları deneyelim!

my_string = "GeCeNin BİR yarısı CİDden örneK bulmak ZoR iş"

print(my_string.capitalize()) # Diğer harfleri de düzeltti! Bu önemli
print(my_string.swapcase()) 
print(my_string.title())
print(my_string.upper())
```

```python
# Burası bayağı önemli, bunu ben de bilmiyordum, öğrenince şaşırdım bayağı bi!
# .lower() ve .casefold() farkı kendini standard ASCII harflerinde değil, bunun dışında belli etmekte

# Bu iki kelimenin aynı olduğunu biliyoruz ama kanıtlamam lazım
word0 = "Straße"
word1 = "STRASSE"

print("lower:", word0.lower(), "casefold:", word0.casefold()) # Casefold ß harfini küçülttü ve ss harfine çevirdi, doğru olanı yaptı
# lower sadece ASCII harfleri küçültebilir, unicode'da sorun yaşamakta

print("lower:", word1.lower(), "casefold:", word1.casefold(), "\n") # Burada sorun yok çünkü hepsi ASCII harfleri

print(word0.lower() == word1.lower()) # Hatalı oluyor doğal olarak
print(word0.casefold() == word1.casefold()) # Doğru cevap alıyoruz

```

#### String Formatlama

Stringler biraz sıkıcı. Çünkü Python'da (ve birçok daha dilde) stringleri editleyemeyiz. `name = "Ahmet"`, `name[3] = "i"` gibi hayaller kurmayalım, maalesef gerçek değiller. Ancak, string'leri dinamik yapabiliriz! Bunun için iki farklı yöntem var! (3. bir yöntem daha mevcut, o C'dekinin aynısı onu atlıyorum o yüzden) Bu yöntemler aracılığı ile değişkenlere göre otomatik olarak ayarlanan string'ler üretebiliriz. Hadi beraber bu iki yöntemi inceleyelim!

***.format()* Metodu:**<br>
Bu metodu sona sakladım. Çünkü en güçlü ve en önemli format bu. Bu metot `"bir şey bir şey {} bir şey {}".format()` şeklinde kullanılır, süslü parantezler içine ise istediğimiz değişkenler gelir. Beraber bu metodu ve bize verdiği inanılmaz güçleri örnek kullanımlar ile inceleyelim:

1. `"'And we'll {}, and we'll {}, and we'll {}' ~Muse, Stockholm Syndrome".format("fly","fall","burn")` örneğinde olduğu gibi süslü parantezlerin içi boş olduğu zaman Python *format()* içinde yer alan verileri sırayla yerleştirir. Unutmayalım ki, her şey bir string'e dönüştürülebildiği için burada int, float vs. verebiliriz!
2. `"'And we'll {luv}, and we'll {het}, and we'll {dye}'" ~Muse, Stockholm Syndrome".format(het="hate", luv="love", dye="d*e")` örneğinde olduğu gibi süslü parantezlerin içinde eğer belli isimler (dictionay 'de olduğu gbi *key* olarak adlandırıyoruz) yer aldığı zaman bunları *format()* içinde değerler (*value*) ile eşleştirebiliriz! Hatta örnekte olduğu gibi sırası farklı bile olabilir!
3. `"'And we'll {1}, and we'll {0}, and we'll {2}'" ~Muse, Stockholm Syndrome".format("hate", "love", "d*e")` örneğinde olduğu gibi süslü parantezlerin içinde indexlere yer verebiliriz. Bu sefer ise *format()* içindeki elemanlara bir listenin öğesi gibi davranırız. Dolayısıyla örnekte olduğu gibi string içinde index'ler sıralı olmak zorunda değildir!

*NOT: Bu örneklerde normal sabit string'ler yer alsa da değişkenler de burada kullanılabilir!*

```python
# Örneksiz olmaz hiç

# Ya İK ya YK veriyorum örnek onu fark ettim az önce
names = ["Ahmet", "Arda", "Selen", "Bilgehan", "Mine", "Ece"]
surnames = ["Birsen", "Çetinkaya", "Yürek", "Turgut", "Dürümle", "Tağluk"] # Soyad sırası ters!

# Tüm adları tek tek gidelim!
for name in names:
    # Tersten gidebilmek için akıllı bir index taktiği yapacağız!
    # İsmin bulunduğu index ile soyadın bulunduğu index'in toplamı 5,
    # O zaman 5 - names.index(name) bana soyadın index'ini verecek!
    surname = surnames[5 - names.index(name)]
    
    print("Hoşgeldin canım İK'm {1} {0}".format(surname, name)) # index mantığı önemli, göstermek için böyle yaptım
```

### Part 2: List, Set, Tuple ve Dict Metodları 

Ders sırasında bazı metotları beraber görmüştük, şimdi bunları tamamlama zamanı! Hemen başlayalım!

#### Ortak Metotlar

Bu metotların hepsi list, set ve dict'lerde bulunmakta. (Tuple'da sadece count() ve index() metotları bulunmakta.)

* `.copy()`: Verilen list, set ya da dict'in bir kopyasını oluşturur. `a = b.copy()` şeklinde kullanılır. Normalde `a = b` kullanımı çoğu zaman sorun olmasa da her zaman en güvenli çözüm `.copy()` kullanmaktır. Ne zaman önem kazandığı henüz öğrenmedik, peki öğrenecek miyiz? İnanın şu anda bilemiyorum :)
* `.clear()`: Tüm elemanları temizler. `list = [1, 2, 3]` ise `list.clear()` işlemi onu boş bir liste yapar.

#### List'e özel metotlar

* `list.count(eleman)`: Verilen elemanın listedeki sayısını döner.
* `list.extend(list)`: Verilen listeyi mevcut listenin sonuna ekler.
* `list.sort()`: Verilen listeyi küçükten büyüğe sıralar. Bunun yerine `sorted()` fonksiyonu daha yaygın kullanılmaktadır, onu da bir sonraki dersimizde öğreneceğiz.

#### Set'e özel metotlar

* `set.pop()`: Verilen set'ten rastgele bir elemanı çıkarır. Niye mesela, değil mi? Çok sorgulamayın Python'ı, yazık edersiniz kendinize.

#### Dict'e özel metotlar

* `dict.popitem()`: En son eklenen *key*-*value* çiftini çıkarır ve döndürür. Döndürmesi detayı önemli, çünkü diğer *pop()* ve benzeri metotlar bir şey döndürmez.
* `dict.setdefault(key, value)`: Eğer *key* zaten dictionary'de varsa onun değerini döndürür. Eğer *key* mevcut değilse de verilen *value* ile beraber dictionary'e ekler.
* `dict.fromkeys(iterable, value)`: Verilen liste (ya da set, tuple) içindeki her bir elemanı *key* olarak alıp, verilen *value* ile eşleyip bir dictionary oluşturur.


```python
# Örnekleme vakti!

random_numbers = [1, 32, 44, 2, 76, 77, 33, 55, 44, 3, 56, 7, 32, 44, 32, 1, 34, 44]

print(f"44 sayısı:{random_numbers.count(44)}") # Kaç tane varmış görelim!

random_numbers = random_numbers.sort()
print(f"sıralanmış hali: {random_numbers}") # Bu da sort'lanmış hali

# Bunu dersten kopyaladım çünkü başka örnek bulamadım
my_dict = {"Başkan": "Erenalp", "İç ilişkiler": "Nisa", "Teknik": "Damla", "Dış ilişkiler": "İzzet"}

print(f"son çift: {my_dict.popitem()}") # Sonuncuyu döndürecek! 
print("zaten varsa:", my_dict.setdefault("Başkan", "Ahmet")) # Başkan zaten var!

print(f"setdefault öncesi: {my_dict}")
my_dict.setdefault("Dış ilişkiler", "Arda")
print(f"setdefault sonrası: {my_dict}")

# Fromkeys anlamak lazım
my_new_dict = dict.fromkeys([1, 2, 3, 4], "Metin Giriniz") # Soldakiler key'ler sağ taraf ise value
print(my_new_dict)
```

### Part 3: Match-Case ve Ternary

Bugüne kadar sadece if-elif-else'den oluşan basit bir hayatımız vardı. Şimdi buna yeni iki müthiş şey ekleme zamanı! Bu iki yeni yöntem ile daha iyi *conditional* yazabiliriz!

#### Match-Case

Eğer başka bir dilde (mesela C) switch-case duyduysanız Python'da match-case de onun kuzeni sayılır. Çalışma mantığı aynı olsa da yapabildikleri (çok acayip şeyler yapabiliyor, onu da sizin keşfinize bırakıyorum.) konusunda çok daha üst bir noktadadır. Match-case bir değişkeni alır ve onun verilen *case*'ler ile eşlemeye çalışır. Bulamazsa *default* varsa onu yapar. Burada case'ler sabit veri olmak zorundadır, değişken bir şey ile direkt yapılamaz. Match-case sadece eşitlik sorgusu yapar!

#### Ternary

Ternary bazen sürekli if-else yazmanızı önlemek amacıyla var olan tek satırlık bir versiyondur. `değer1 if condition else değer2` şeklinde yazılır ve sorgu *True* ise ilk değer değilse ikinci değer eşlenir. Böylece basit if-else'leri uzun uzun yazmamış ve kodumuzu okunur kılmış oluruz.

```python
# Örneklerle yola devam!

my_num = 267

# my_num'ı deneyecek, evet bu kadar örneklerim bitti
match my_num:
    case 264:
        print(264)
    case 265:
            print(265)
    case 266:
            print(268)
    case 267:
            print(267) 
    case 268:
            print(268) 
    case _:
            print(269)
```

```python
# Ternary'ler favorim ya

my_num = 267

my_name = "Ahmet" if my_num <= 300 else "Arda"
# Cidden örneklerim bu saatte (1.00) çok kötü :(

print(my_name)
```
