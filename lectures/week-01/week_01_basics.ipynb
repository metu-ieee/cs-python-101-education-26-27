### Part 2: Değişkenler ve *Primitive* (Temel) Veri Tipleri
Her şeyden önce bizim bir şeyleri bir şekilde bir yerde tutmamız lazım. Bunun için ise *variable* (değişkenler) imdadımıza yetişiyor!

#### Değişken Nedir?
Değişkenler bizim kodlama sırasında kullandığımız en temel parça, hatta her şeyin bir başlangıcı. Değişkenler ile bir veriyi kendine ait bir yerde saklayıp ihtiyacımız olduğu yerde kullanmayı sağlar. Değişkenlerde tuttuğumuz veri üzerinde çok çeşitli işlemler yapabiliriz. Güzel bir örnek:
```python
name = "Ahmet"
num = 1
```
Burada **"name"** ve **"num"** iki farklı değişken. Bir tanesi bir *string* taşırken diğeri bir sayı taşıyor. İki değişken farklı **veri tipine** sahip.

#### Veri Tipi Nedir?
Yukarıda da olduğu gibi, değişkenler farklı türlere sahip. Programlama dillerinde **veri tipleri** ikiye ayrılır: *primitive* (temel) ve *complex/composite* (karmaşık/bütünleşik) veri tipleri. Bu hafta sadece temel veri tiplerini öğreneceğiz merak etme! Temel veri tipleri oldukça kolay! İşte Python'daki 4 temel veri tipi:
* **Strings (str):** Tek veya çift tırnak işareti içinde yazılan metin öğeleri.
* **Integers (int):** Tam sayılar.
* **Floating Points (float):** Ondalık sayılar.
* **Booleans (bool):** *True* (doğru) ve *False* (yanlış) ikili verisi. (Büyük harfe dikkat!)

````python
# Hadi biraz daha örnek görelim!
name = "Walter White" # Bu bir string!
nickname = 'Heisenberg' # Bu da bir string!
# Evet hem "çif tırnak" hem de 'tek tırnak' tamamen aynı!
# Arada hiçbir fark yok (Python için geçerli).

cash = 100000 # Bu bir integer!
age = 0b1000110 # Bu da bir integer! Ama binary.
# 0b ile binary (ikili), 0o ile octal (sekizli) ve 0x ile hexadecimal (onaltılı) ile yazabilirsiniz!

rating = 9.5 # Bu bir float!
pizza = 6.022e23 # Bu da bir float! Ama bilimsel gösterim.
# e (ya da E) ile üslü sayı gösterebilirsiniz!
# 6.022e23 = 6.022 x 10^23

jesse = 1_234_567 # Sayı uzunsa alt tire kullanabilirsin! Python onu yok sayar.
# Alt tireyi hem integer hem de float için kullanabilirsin.
half = .5 # Python bunu 0.5 algılar.

rich = True # Bu bir bool!
good = False # Bu da bir bool! 
# Büyük harfe dikkat!

# Bunu yapmak zorundaydım :P
print("Now, say my name.")
print(nickname)
print("You're goddamn right.")

# Boolean kontroller böyle yapabilirsiniz!
print(0b1000110 == 70 == 0o106 == 0x46)
# Basit matematik de yapabilirsiniz!
print(pizza / rating)
print(jesse + half)
````

#### Değişken Nasıl İsimlendirilmez?
Python biraz dikkatli bir dildir. Dolayısıyla bir takım kuralları bulunmakta, öyle her değişkeni istediğimiz gibi isimlendiremiyoruz maalesef. Kurallar şu şekilde:
* Türkçe karakter kullanımı serbest. Evet `söğüş = "leziz"` yazarsanız kod çalışmaya devam eder. Ama kodlarda (özellikle [PEP 8](https://peps.python.org/pep-0008/) kuralları gereği) çok tavsiye edilmez, değişkenlerin isimlerini İngilizce tutmak en iyisi.
* Değişkenler sayı ile başlayamaz, sadece harf ve alt tire ile başlayabilir. `_iskender = 1.5` doğruyken `1skender = 1.5` hata verir.
* Değişkenlerde boşluk ya da "-", "+", ve "$" gibi özel işaretler bulunamaz. `kağıt_kebabı = "iyi pişmiş"` doğru çalışır, ancak `kağıt kebabı+ = "iyi pişmiş"` yazarak israf edersiniz.
* Python'da önümüzdeki haftalarda öğreneceğimiz "and", "for", "while", "elif" ve "if" gibi kelimeler değişken olamaz. `ayşe = True` çalışırken `elif = False` maalesef sorun çıkarır. (Elif'ler üzgün)
* Büyük harf - Küçük harf farklıdır. `Baklava = 400` ile `baklava = 400` farklı değişkenlerdir.

Normalde dünyadaki çoğu yazılımcı Python'ın [PEP 8](https://peps.python.org/pep-0008/) yazım standardına uygun bir şekilde kod yazarlar. Bu standart, yukarıda olan kurallara yenilerini ekler.

#### Type() Nedir?
Evet 4 tane tür öğrendik, ama ben bunun türünü bilmiyorum! Sakin ol, Python'ı yazan kişiler bunu öngördü ve **`type()`** diye çok güzel bir araç yazdılar. Bu araç ile her değişkenin türünü anında öğrenebilirsin!

````python
# Hadi hemen deneyelim!
Doctor = "House"
its_never = 100
Lupus = True

# Değişkene atayalak bakalım
doc = type(Doctor)
print(doc)
# What if?
print(type(doc))

# Değişkensiz de olur
print(type(its_never))

# Kıyas vakti!
print(type(Lupus) != type(Doctor))
print(type(its_never) == type(.2))
````

### Part 4: *Type Casting* (Veri Tipi Değiştirme)
`input()` fonksiyonunun her zaman bir *string* döndürdüğündem bahsetmiştik. Peki bu durumda ne yapacağız? Çözümü oldukça basit: **Type Casting** ile dinamik bir şekilde veri tipini değiştirmek! **Type Casting** aşağıdaki fonskiyonlar ile yapılabilir:
* `int()` ile bir veriyi integer'a dönüştürebilirsiniz.
* `float()` ise ondalık sayı yapar.
* `str()` de her şeyi bir string yapar.
* `bool()` ise bambaşka bir hikaye.

````python
# Float'tan integer'a dönüştürelim
my_flt = 7.77
print(type(my_flt).__name__) # __name__ ile sadece 'float' yazar
my_flt = int(my_flt)
print(type(my_flt).__name__)
print(my_flt)

# Peki tam tersi?
my_int = 11
print(type(float(my_int)).__name__) # my_int değişmedi, sadece içindeki sayıyı alıyoruz
print(type(my_int).__name__)
print(float(my_int))

# String dönüşümleri
print(str(my_int)) # tırnaklar yok inanmıyorum
print(type(str(my_int)).__name__)
# Tersi çok tehlikeli
my_flt = str(my_flt)
print(type(my_flt).__name__)
print(int(my_flt))
my_flt = "Gustavo Fring"
print(type(my_flt).__name__)
# print(int(my_flt)) # Hata verecek!

# Bool biraz korkunç
print(type(bool(my_int)).__name__)
print(bool(my_int))
print(bool(my_flt))
print(bool(""))
print(bool(0))
# Kural basit:
# Boş ya da 0 ise False
# Değilse True
````
