### Part 4: *Type Casting* (Veri Tipi Değiştirme)
`input()` fonksiyonunun her zaman bir *string* döndürdüğündem bahsetmiştik. Peki bu durumda ne yapacağız? Çözümü oldukça basit: **Type Casting** ile dinamik bir şekilde veri tipini değiştirmek! **Type Casting** farklı veri türlerini birbirine dönüştürmeyi sağlar. **Type Casting** aşağıdaki fonskiyonlar ile yapılabilir:
* `int()` ile bir veriyi integer'a dönüştürebilirsiniz.
* `float()` ise ondalık sayı yapar.
* `str()` de her şeyi bir string yapar.
* `bool()` ise bambaşka bir hikaye. Az sonra aşağıda göreceğiz.

````python
# Bir float ile başlayalım:
my_flt = 7.77
print(type(my_flt)) # Önce bi türünü doğrulayalım!

# Şimdi bu float'ı bir int yapalım!
my_flt = int(my_flt) # Burada my_flt'yi bir integer yapıp tekrar my_flt'ye eşitledim

print(type(my_flt)) # Artık bana int demesi lazım
print(my_flt) # .'dan sonrası direkt yok oldu! 
# Python float -> int dönüşümlerinde her zaman noktadan sonrasını direkt yok eder!
````

````python
# Bir integer ile başlayalım:
my_int = 11

# Farklı bir şey deneyerek başlayalım
print(type(float(my_int))) # Dikkat: Burada my_int'i değiştirmiyorum, my_int'i bir float yapıp type()'a yolluyorum
print(type(my_int)) # Hadi my_int'in değişmediğini kanıtlayalım!
# my_int değişmedi çünkü my_int'i hiçbir yerde = kullanıp değiştirmedim, my_int'i değiştirmek için ona bir şey atamam gerekliydi!

# Şimdi int -> float dönüşümüne bakalım:
print(float(my_int)) # Python sonuna ".0" koydu, çünkü bizim onun bir int değil de bir float olduğunu fark etmemizi istiyor!
````

````python
# Hadi string'lere bakalım!
print(str(my_int)) # Python tırnaklarla yazdırmadı, acaba gerçekten dönüştürdü mü?
print(type(str(my_int))) # Evet gerçekten de dönüştürmüş

# float -> str ve int -> str dönüşümleri oldukça sade, direkt bir string oluyorlar.
# Ancak str'den olan dönüşümlerde dikkatli olmalıyız

# Örnek 1:
print("--1. Senaryo--")
my_str = "20"
print(float(my_str)) # Sorunsuz çalışıyor! Python sonuna ".0" bile koydu!
print(int(my_str)) # Bu da sorunsuz!
# Özetle Python tam sayıdan oluşan bir string'i rahatça çevirdi

# Örnek 2:
print("--2. Senaryo--")
my_str = "20.6"
print(float(my_str)) # Sorunsuz çalışıyor! Python bunu rahatlıkla tanıdı!
#print(int(my_str)) # Python bunu sevmedi, 20.6 bir tam sayı olmadığı için burada hata verdi! Yorumu kaldırıp deneyebilirsiniz!
# Özetle Python bir ondalık sayı halindeki string'i float'a çevirebildi ama int'e çeviremedi.

# Örnek 3:
print("--3. Senaryo--")
my_str = "   20   " # Başına ve sonuna birkaç boşluk koyduk
print(float(my_str)) # Sorunsuz çalışıyor! Python sonuna ".0" bile koydu!
print(int(my_str)) # Bu da sorunsuz!
# Özetle Python string başındaki ve sonundaki boşlukları yok saydı

# Örnek 4:
print("--4. Senaryo--")
my_str = "20a" # Başına ve sonuna birkaç boşluk koyduk
#print(float(my_str)) # Python burada a'yı görünce bunu tanıyamadı
#print(int(my_str)) # Python burada a'yı görünce bunu tanıyamadı

# Özetle Python string içerisinde saf halde tek bir sayı görmediği zaman tanıyamıyor,
# Bunun yanı sıra normal string'leri int veya float'a çevirmek de mümkün değil,
# str -> float ve str -> int dönüşümleri için sahip olduğumuz string tek parça halinde bir sayı olmalı!
````

````python
# Boolean inceleme vakti!
print(type(bool(my_int))) # Önce dönüştürüyor mu onu kanıtlayalım!

print("--sayılar ve bool--")
my_int = 156 # 0'dan büyük bir sayı yapalım
print(bool(my_int)) # Buna True dedi!

my_int = -156 # Negatif bir sayı deneyelim
print(bool(my_int)) # E gene True dedi, ne zaman False diyecek?

my_int = 0 # 0'ı deneyelim artık
print(bool(my_int)) # Sonunda False dedi!

# Özetle int -> bool ve float -> bool dönüşümünde:
# 0 ise False değilse True

print("--string ve bool--")
my_str = "Gustavo Fring"

print(bool(my_str)) # Buna True dedi!

# Boş string deneyelim:
print(bool("")) # Buna False dedi!

# Önemli:
print(bool(" ")) # Boşluk karakteri Python tarafından algılanır, bu bir boş string değil, True dönecek!
print(bool("0")) # İçindeki 0 direkt algılanmaz, string boş değil! True dönecek.

# Özetle, str -> bool dönüşümlerinde:
# Eğer string boş ise False, dolu ise (boşluk karakteri dahil!) True
````

````python
# Bool'dan öbür yöne dönüş nasıl?
print(int(True)) # True her zaman 1 olarak döner!
print(float(False)) # False ise her zaman 0 olarak döner!

print(str(True)) # bool -> str dönüşümlerinde ise direkt isimleri bir string olur

# NOT: Boolean aslında özel bir integer türü! Bu da demek oluyor ki Matematiksel işlemler sırasında True ve False kullanabiliriz!
# Buna dair örnekleri aşağıda göreceğiz.
````

### Part 5: Basit Aritmetik ve Karşılaştırma
Aritmetik kodlama için de kaçınılmaz, her kodun bir kısmında illaki basit aritmetik işlemlerine rastlamanız mümkün. Dolayısıyla kağıt üzerinde yaptığımız herşeyi kodda da yapabilmeyi öğrenmemiz lazım. Hadi, güzel bir başlangıç yapalım:
* **Toplama ve Çıkarma:** Bunlar en basitleri ve hiçbir fark yok. "+" ve "-" işaretleri toplama ve çıkarmayı ifade ediyor.
* **Çarpa ve Bölme:** Bunlarda çok ufak farklar bulunmakta. "*" ve "/" işaretleri sırasıyla çarpma ve bölme için kullanılıyor.
* **Üs Alma:** Üs alma ise normalde yaptığımızdan çok farklı. "**" işareti ile üs almaktayız. Örneğin `5**2 = 25`.
* **Mod Alma** Mod alma işlemi matematikte bir sayının diğerinden verdiği kalanı ifade eder. Python'da "%" sembolü mod almak içindir. Örneğin `17%4 = 1`, bu ifade kağıt üzerinde 17 ≡ 1 (mod 4) olarak ifade edilir.
* **Kalansız Bölme:** Normal bölme işlemi her zaman bir *floating point* döndürür. *Integer Division / Floor Division* işlemi ise kalansız bir sonuç almamızı sağlar. Dönen değer bir *integer* olur. Python'da "//" sembolü ile yapılır. Örneğin `17//4 = 4`.

Bu matematiksel operatörlerin yanı sıra karşılaştırma yapmanızı sağlayan bazı operatörler bulunmakta, **bunlar her zaman boolean döndürürler**:
* **Büyüktür ve Küçüktür:** Bunlar en bariz olanları. ">" büyüktür, "<" ise küçüktür sembolü.
* **Büyük Eşit ve Küçük Eşit:** Bunlarda ise iki sembol beraber. ">=" ve "<=" ile büyük eşit ve küçük eşiti ifade edebilirsiniz. Dikkat edin, eşittir sembolü her zaman sağ tarafta!
* **Eşit Mi:** "==" sembolü ile iki şeyin eşit olup olmadığını sorgulayabiliriz. Örneğin `17.3 == 1.73e1` ifadesi True, `"Michael Jackson" == "Michael Jordan"` ifadesi False döndürür.
* **Eşit Değil Mi:** Bu ise üsttekinin tamamen tersi. "!=" ile ifade edilir. Yukarıdaki örneklerde "==" yerine "!=" koyarsanız tam tersi sonuçlar elde edersiniz.

````python
# Hadi biraz örneklerle deneyelim!

print("--toplama/çıkarma--")
print(10 + 5) # Standart toplama işlemi
print(10 - 5) # Standart çıkarma işlemi

print("--çarpma/bölme--")
print(10.5 * 7.5) # Standart çarpma işlemi
print(10.5 / 7.5) # Standart bölme işlemi

print("--üs alma/mod alma--")
print(5**3) # Standart üs alma işlemi
print(17%4) # Standart mod alma işlemi

print("--bölme vs. kalansız bölme--")
# Buraya dikkat edelim!
print(10/2) # Python tam bölündüğü halde bir int yerine bir float verdi!
# Çünkü Python standart bölme işlemini gördüğü gibi int ve bool'u float'a çevirir!

# Bunu engellemek için kalansız bölme kullanalım:
print(10//2) # Artık int dönecek!
print(17//4) # Burada ise sonucu int'e çeivirip verdi!

# Kısayol: int(x/y) = x//y
````

````python
# Karşılaştırma vakti!

print("--büyük/küçük--")
print(10 > 5) # Standart büyüktür sorgusu
print(10 < 5) # Standart küçüktür sorgusu

print("--büyük/küçük eşit--")
print(10 >= 10) # Standart büyük eşittir sorgusu
print(10.5 <= 10) # Standart büyük eşittir sorgusu
# Unutmayalım! önce < ya da > sonra =, ve bitişik!

print("--eşit mi değil mi--")
print(17.4 == 1.74e1) # Standart eşit mi sorgusu
print(10.0 != 10) # Standart eşit değil mi sorgusu, Python burada gerekli int -> float dönüşümünü yaptı ve kıyasladı!
````

````python
# Ufak Not: Python matematiksel işlemler sırasında işleme göre otomatik olarak bool -> int -> float dönüşümlerini yapar!
print(6 * True) # True -> 1
print(False / 7) # False -> 0 -> 0.0

print(10.5 * 4) # sonuç bir tam sayı olsa da Python çoktan 4'ü bir float yaptı bile!
````

### Part 6: *Conditionals* (Koşul İfadeleri) ve *Logical* (Mantıksal) Operasyonlar
Tamam, güzel artık veri girdimiz ve çıktımız var ve veri türlerini değiştirebiliyoruz. Ancak kodumuz sabit, durumdan duruma farklı tepki veremiyoruz. Bir şeyler eksik... Bize koşul ifadeleri ve mantıksal operasyonlar lazım!

#### Conditionals Nedir? 
Kodumuzda çeşitli koşulları ve olasılıkları karşılamak için bir kontrol akışı oluşturmamız gerekmekte. Kontrol akışı, verinin sorgulandığı, durumlarının tek tek incelendiği ve karşıladığı durumlar uyarınca birtakım işlemlerin gerçekleştiği bir akıştır. Bunun için bazı ifadelere ihtiyacımız bulunmakta:
* **`if`**: Kontrol akışını başlatmak için kullanılır. İlk soru burada sorulur, `if` bize bir *boolean* ile dönüş yapar. Doğru ise hemen içindeki kod bloğu çalıştılır, yoksa atlanır.
* **`else`**: `if` kontrolü *False* döndüğü zaman altında bir else bloğu varsa o bloğun içindeki kod çalıştırılır. 

**Bir dakika! Ya birden fazla kontrol etmem gereken şart varsa?**
O zaman `elif` kullanıyoruz! (Arkadaşın olan değil, öbür elif bu):
* **`elif`**: *else* ve *if* ifadelerinin birleşimi ile oluşur. Üstünde yer alan `if` bloğu *False* dönerse bu blok sorgulanır. Birden fazla `elif` alt alta kullanılabilir!

````python
# Hadi biraz deneyelim!
my_name = input("Knock knock! Who are you?")
if(my_name == "Erenalp"): # if ve elif için parantez gerekli!
    print("Ooooo hoşşgeldin başkanımm!")
elif(my_name == "Arda"):
    print("CS'in kralı gelsin")
else:
    print("Seni tanımıyorum git buradan!")

# Normal bir kontrol akışında if, elif ve else beraber bulunur!
# Ancak bu zorunlu değil!
# Tek başına bir if kullanılabilir!
# Sadece bir else ve if olabilir!
# Sadece bir if ve birkaç elif olabilir!
# Zorunlu olan tek şey bir if ile başlaması gerektiği (if ortada olmaz, o zaman kontrol akışları ayrılmış olur) ve else'in varsa en sonda olması gerektiği
````

````python
# Python if else akışlarını iç içe koymamıza imkan tanır!
# Hadi beraber bir asal sayı tespit mekanizması yapalım!

my_num = int(input())
if(my_num % 2 == 0): # Çift mi değil mi bakalım, çift ise eleyelim.
    print("Asal Değil")
else:
    if(my_num % 3 == 0):
        print("Asal Değil")
    else:
        if(my_num % 5 == 0):
            print("Asal Değil")
        else:
            print("Belki de asal") # Devamına üşendim

# Elbette bu asal sayıları bulmak için kötü bir yöntem, burada sadece kontrol akışlarının iç içe geçebileceğini göstermek istedim.
````

#### Logical Operasyonlar Nelerdir?

Şu anda her bir sorguda aynı anda tek bir şeyi kontrol edebiliyoruz, peki birden fazla şeyi aynı anda kontrol etmem gerekiyorsa, ya da bir koşulu sağlamak için 5 şarttan en az birisi yeterliyse ne yapmam lazım? O zaman da mantıksal operasyonlar yardımımıza yetişiyor! Hemen tanıyalım:
* **`and`**: Sadece iki tarafı da doğruysa *True* döner, iki yanından birinin yanlış olması *False* dönmesi için yeterlidir!
* **`or`**: İki tarafında yer alan ifadelerden birisinin doğru olması *True* dönmesi için yeterlidir. *False* dönmesi için iki tarafının da yanlış olması gereklidir!
* **`not`**: Kendinden sonra gelen ifadenin tersini alır.

````python
# Şimdi ise beraber bir XOR GATE yapalım! XOR (Exlusive OR) GATE kendisine verilen iki sayı aynı ise 0, farklı ise 1 verir, yani:
# A | B | Q
# 0 | 0 | 0
# 0 | 1 | 1
# 1 | 0 | 0
# 1 | 1 | 1

# Yardımcı bilgi: XOR açılımı (A AND B') OR (A' AND B) şeklindedir, hadi bunu kodlayalım:
my_num = 1
my_num2 = 0

xor = (my_num and not my_num2) or (not my_num and my_num2)

print(xor) # Peki neden my_num 1 iken True ve False dönüyor da 0 iken 1 ve 0 dönüyor? Açıklaması "Meraklısına" kısmında!
````
