### Part 1: Basit Aritmetik ve Karşılaştırma
Aritmetik kodlama için de kaçınılmaz, her kodun bir kısmında illaki basit aritmetik işlemlerine rastlamanız mümkün. Dolayısıyla kağıt üzerinde yaptığımız herşeyi kodda da yapabilmeyi öğrenmemiz lazım. Hadi, güzel bir başlangıç yapalım:
* **Toplama ve Çıkarma:** Bunlar en basitleri ve hiçbir fark yok. "+" ve "-" işaretleri toplama ve çıkarmayı ifade ediyor.
* **Çarpa ve Bölme:** Bunlarda çok ufak farklar bulunmakta. "*" ve "/" işaretleri sırasıyla çarpma ve bölme için kullanılıyor.
* **Üs Alma:** Üs alma ise normalde yaptığımızdan çok farklı. "**" işareti ile üs almaktayız. Örneğin `5**2 = 25`.
* **Mod Alma** Mod alma işlemi matematikte bir sayının diğerinden verdiği kalanı ifade eder. Python'da "%" sembolü mod almak içindir. Örneğin `17%4 = 1`, bu ifade kağıt üzerinde 17 ≡ 1 (mod 4) olarak ifade edilir.
* **Kalansız Bölme:** Normal bölme işlemi her zaman bir *floating point* döndürür. *Integer Division / Floor Division* işlemi ise kalansız bir sonuç almamızı sağlar. Dönen değer bir *integer* olur. Python'da "//" sembolü ile yapılır. Örneğin `17//4 = 4`.

Bu matematiksel operatörlerin yanı sıra karşılaştırma yapmanızı sağlayan bazı operatörler bulunmakta, **bunlar her zaman boolean döndürürler**:
* **Büyüktür ve Küçüktür:** Bunlar en bariz olanları. ">" büyüktür, "<" ise küçüktür sembolü.
* **Büyük Eşit ve Küçük Eşit:** Bunlarda ise iki sembol beraber. ">=" ve "<=" ile büyük eşit ve küçük eşiti ifade edebilirsiniz. Dikkat edin, eşittir sembolü her zaman sağ tarafta!
* **Eşit Mi:** "==" sembolü ile iki şeyin eşit olup olmadığını sorgulayabiliriz. Örneğin `17.3 == 1.73e1` ifadesi True, `"Michael Jackson" == "Michael Jordan"` ifadesi False döndürür.
* **Eşit Değil Mi:** Bu ise üsttekinin tamamen tersi. "!=" ile ifade edilir. Yukarıdaki örneklerde "==" yerine "!=" koyarsanız tam tersi sonuçlar elde edersiniz.

````python
# Hadi biraz örneklerle deneyelim!

# Para
bank = "Ziraat Bankası"
Kyk = 4000
dad_money = 13063.42

# Ne satın alıyorum, evet bunlarla besleniyorum
coffee = 180
pop_kek = 14.7
grain_chips = 20

day_length = 24
month_length = 30 * day_length

# Bu işlemleri niye yapıyorum bilmiyorum 
print(coffee % day_length)
print(coffee / day_length)
print(coffee // day_length)
print(type(coffee // day_length).__name__) # Evet integer!
print(pop_kek**3) # Evet neden pop kekin üssünü almayalım ki!
print(day_length - 4 == grain_chips)
print(int(pop_kek) != pop_kek)
print(str(pop_kek) != pop_kek) # Çünkü farklı veri tipleri

# Manalı hesapların vakti
monthly_income = Kyk + dad_money
daily_income = monthly_income / 30
daily_spend = 3 * coffee + 3 * pop_kek + grain_chips # Evet, artiemtik operatörleri kağıt üzerinde olduğuyla aynı işlem önceliğine sahip!
am_i_broke = daily_spend >= daily_income
am_i_rich = monthly_income > daily_spend * 40 # Kırk günlük masrafımdan fazla kazanıyorsam zenginimdir
money_left = monthly_income - 24 * daily_spend # Ayda 24 gün harcama yaptığımı varsaydım
coffee_count = monthly_income // coffee # Ayda kaç kahve alabildiğim önemli bir metrik

print("--Para Analizi--")
print("Günlük Gelir:", daily_income)
print("Fakir Miyim:", am_i_broke)
print("Zengin Miyim:", am_i_rich)
print("Kalan Para 2500'den Küçük ya da Eşit Mi:", money_left <= 2500)
print("Bir Ayda Alınabilen Kahve Sayısı:", coffee_count)

````

### Part 2: Başka Veri Tipleri ve Dönüşümleri
Evet daha önce bazı veri tiplerini görmüştük. Bunlar içerisinde *binary*, *octal*, ve *hexadecimal* gösterimler de mevcuttu. Şimdi hem bunlara olan dönüşümleri hem de yeni bir veri tipini öğrenme vakti geldi!

#### Complex Veri Tipi
Lisede kompleks/imajiner sayıları görmüştük (Yani ben gördüm ve o kadar yaşlı değilim!). Bu sayılar normalde "a + bi" olarak ifade ediliyordu. Python'da ise bunu "a + bj" olarak ifade edebiliyoruz!
Örneğin `5 - 3j` ifadesi Python'da geçerli olup türü de complex olmakta.

#### Binary, Octal ve Hexadecimal dönüşümler
Elinizde bir sayı varsa ve bunu bir sebepten ötürü bu biçimlerden birine dönüştürmek istiyorsanız yapması çok basit! Her biri için şu şekilde iki yönlü dönüşümler yapabilirsiniz:
* **Binary:** `bin()` fonksiyonu içine aldığı sayıyı binary yapıp string olarak döndürür. Geri almak için çıktıyı int() içinde `int(çıktı, 2)` şeklinde kullanabilirsiniz.
* **Octal:** `oct()` fonksiyonu içine aldığı sayıyı octal yapıp string olarak döndürür. Geri almak için çıktıyı int() içinde `int(çıktı, 8)` şeklinde kullanabilirsiniz.
* **Hexadecimal:** `hex()` fonksiyonu içine aldığı sayıyı hex yapıp string olarak döndürür. Geri almak için çıktıyı int() içinde `int(çıktı, 16)` şeklinde kullanabilirsiniz.

````python
my_int = 27
my_int_bin = bin(my_int)
my_int_oct = oct(my_int)
my_int_hex = hex(my_int) 
# Bu variable'ları biribirine input olarak veremiyoruz
# Çünkü artık hepsi bir string

# Deneyelim!
print("--Sayılar--")
print(my_int_bin)
print(my_int_oct)
print(my_int_hex)
print(type(my_int_bin).__name__) # Evet bir string!
print(type(my_int_bin) == type(my_int_oct) == type(my_int_hex)) # Hepsi bir string!
# Hepsinin başında yer alan 0b, 0o ve 0x bizim birbirine input olarak vermemizi engelliyor

# Float denesek mi?
my_flt = 10.44
# my_flt_bin = bin(my_flt) # Bu satır hata veriyor!
# print(my_flt_bin) # Çünkü input bir integer olmak zorunda! Yorumları kaldırıp deneyebilirsin!

# Geri çevirmeye çalışalım:
my_int_1 = int(my_int_bin, 2)
my_int_2 = int(my_int_oct, 8)
my_int_3 = int(my_int_hex, 16)

# Deneyelim!
print("--Düzeltilmiş Sayılar--")
print(my_int_1) # Oldu!
print(my_int_2)
print(my_int_3)
print(type(my_int_1).__name__) # Evet bir integer!
print(type(my_int_1) == type(my_int_2) == type(my_int_3)) # Hepsi bir integer!

# Peki ?..
# my_int_4 = int(my_int_bin, 8) # Hata verdi! Demek ki, o sayı bizim decimal'a döndürdüğümüz base olmalı!
# my_int_5 = int(my_int_bin, 2, 8) # Hata verdi! Demek ki, tek seferde sadece decimal'a dönüştürebiliyoruz!
my_int_6 = int("66",7) # Base 7'yi deneyelim! 66, base 7'de 48 demek
print(my_int_6) # Oldu! Demek ki Python diğer number base'leri destekliyor.
# NOT: Pyton Base 36'ya kadar (dahil) desteklemekte
# NOT: Binary, Octal ve Hexadecimal dışında ek bulunmadan direkt string almakta!
````

````python
# Şimdi de complex'i deneyelim
my_comp = 5 - 7j
print(type(my_comp).__name__) # Evet complex dedi!
# Unutmayın ki, j kullanmanız gerekiyor!

# complex() fonksiyonunu kullanalım!
my_int = 11
print(complex(my_int)) # +0j ekledi!

# Tersi yönde neler oluyor?
# print(int(my_comp)) # Kabul etmedi! int() string kabul eder, complex değil!

# Aritmetik İşlemler?
my_comp_1 = complex(5,7) # Evet, böyle de oluşturabiliyoruz!

print("--Aritmetik--")
print(my_comp + my_comp_1)
print(my_comp - my_comp_1)
print(my_comp * my_comp_1)
print(my_comp / my_comp_1) # Bunun sonucu nasıl açıklayacağımı bilemedim
# Okulda boşuna komplex sayıları bölmezsiniz dememişler

# Birkaç şey daha
print(my_comp.real) # Sadece reel kısmı
print(type(my_comp.real).__name__) # Float oldu! Çünkü örneğim 5.42 + 11.3j de olabilirdi, Python hazırlıklı!
print(my_comp.imag) # Sadece imajiner kısım
print(type(my_comp.imag).__name__) # Float oldu! Çünkü örneğim 5.42 + 11.3j de olabilirdi, Python hazırlıklı!
print(abs(complex(3,-4))) # Büyüklüğü yani "a + bi" (ya da Python a + bj) için kök(a**2 + b**2) işlemi. Evet, yine float!
````

### Part 3: Hayat Kurtarıcı `print()` Bilgileri
Evet, hem `input()` hem de `print()` fonksiyonlarını gördük. Girdi alma konusunda önümüzdeki haftalarda yeni şeyler öğreneceğiz. Çıktı konusunda ise fonksiyonlar haftasında çok daha fazlasını öğrenecek olmamıza rağmen şimdiden öğrenebileceğiniz hayat kurtarıcı bilgiler bulunmakta. Hadi başlayalım:
* `print()` fonksiyonu'nda "," kullanarak sınırsız sayıda şeyi tek seferde çıktı verebilirsiniz. Örneğin `print("aa", b, count, "14.5", 3e7, 0xdd, 2)` ifadesi hepsini beraber çıktı olarak verir ve hepsinin arasına bir boşluk koyar.

### Part 4: Önemli String Metotları
String'leri az biraz kullanmayı öğrendik ama hayatınızı kurtaracak çok daha fazlası var! Bu hafta en önemlilerini öğreneceğiz, daha sonra daha gelişmiş yöntemleri göreceğiz. Python'da String yöntemleri çok önemli, o yüzden önümüzdeki haftalarda detaylara ineceğiz!

#### String Nasıl Formatlanır?
Formatlama mı? Ne? Evet, Python'da string'ler formatlanabilir! String'lere dinamik olarak bir şeyler eklemek istiyorsak onları formatlamamız lazım. Python'da 3 tane yöntem var, ancak biz şimdi 2 tanesini basit olarak göreceğiz.
1. **F-String:** *formatted string* yani f-string çok fazla yerde kullandığımız bir formatlama yöntemi. String'in başına bir "f" (ya da F) koyarak elde edebilirsin! Sonrasında ise {} içine koyacağın her şeyi bir değişken ya da işlem ile ilişkilendirebilirsin! Örneğin `f"JoJo's Bizarre Adventure Part {number}: {part_name}"` örneğinde Python `number` ve `part_name` değişkenlerini bulup yerleştirir!
2. **.format():** *.format()* metodu bir stringi *f-string* yöntemine benzer şekilde format etmeye yarar. Index kullanımı gibi çeşitli özellikleri destekler. En büyük farkı değişkenlerin *.format()* içinde tanımlanabilir olmasıdır. Örneğin `"JoJo's Bizarre Adventure Part {number}: {part_name}".format(number=5, part_name="Golden Wind")` örneğinde Python `number` ve `part_name` değişkenlerini otomatik yerleştirir!

````python
# Hadi Deneyelim!
print("JoJo's Bizarre Adventure Part {number}: {part_name}".format(number=5, part_name="Golden Wind"))

# Tek bir tane olunca boş {} ile direkt kullanabilirsin, değişken tanımlamana da gerek yok!
print("But it was me, {}".format("Dio!"))

# Index kullanabilirsin (ne olduğunu bir sonraki hafta öğreneceğiz)
print("AESS's stand is {1}, CS's stand is {0}".format("Baby Face", "Aerosmith"))

peak = "JoJo's "
anime = "Bizarre Adventure"

print(f"I watch {peak + anime} btw.")
````

#### Escape Character Nedir?
Bir string içinde yer alan "\" karakteri *escape character* olarak algılanır. Escape character Python tarafından çeşitli şekillerde kullanılır. Hadi bazılarını keşfedelim:
* **Yeni Satır:** `\n` kombinasyonu string'de yeni satıra geçmeyi sağlar. Örneği `print("Kardeşler\nPide\nFırını")` ifadesinde her kelime bitiminde Python bir alt satıra geçip yeni kelimeyi yazar.
* **Backslash Yazma:** "\" tek başına yazıldığında python onu yazmayacağı için `\\` şeklinde yazmamız gerekir. Örneğin `print("Büyük\\Ayran")` ifadesi Büyük\Ayran çıktısı verir.
* **Tırnak İşaret:** \" ve \' kombinasyonları string içinde yazımına imkan verir, string'i bozmaz. Örneğin `print("1.5 \"İskender\"")` ifadesi ekrana 1.5 "İskender" yazar.
**ÖNEMLİ NOT:** Farklı tırnak işaretleri birbirinin içinde kullanılınca gerek yoktur. "'a'" ekrana 'a', '"a"' ekrana "a" yazdırır. "\" kullanmaya gerek kalmaz.
* **R-String:** *Raw String* string içindeki tüm \ karakterlerini `\\` kullanmadan yazdırmaya yarar. Örneğin `print(r"C:\Users\name")` ifadesi C:\Users\name yazdırır.

````python
# Deneme vakti!
print("Kardeşler\nPide\nFırını")
print("Abi bana bi\\Büyük Ayran")
print("1.5 \"İskender\" bana da abi")
print("1.5 'İskender' bana da abi")
print(r"C:\Sadece\İskender\Düşünebiliyorum")
````

#### String Metotları
String'leri düzenlemek için Python'da birçok metot bulunmakta. Bu metotları string'in sonuna "." koyarak ekleyebilirsin! Hadi önemli olanları inceleyelim:
* `lower()`: Tüm string'i küçük harflere çevirir.
* `upper()`: Tüm string'i büyük harflere çevirir.
* `title()`: Tüm string'deki kelimelerin hepsinin ilk karakterini büyük yapar.
* `capitalize()`: Tüm string'deki sadece ilk karakteri büyük harf yapar.
* `replace("eski","yeni",sayı)`: String'de yer alan "eski"leri *sayı* kadar "yeni"ye çevirir. *sayı* verilmezse hepsini yapar.
* `split("ayraç",sayı)`: String'i *sayı* kadar "ayraç"tan ayırıp *sayı*+1 kadar string elde eder. *sayı* yoksa her "ayraç"tan ayırır. "ayraç" yoksa boşluklardan böler.
* `count("terim")`: Tüm string'de kaç tane "terim" olduğunu bulur.
* `strip()`: String'in başı ve sonundaki boşlukları siler.

Ayrıca, `len()` fonksiyonu içine aldığı string'in uzunluğunu verir!

````python
# Hadi deneyelim!

manifest = "iŞte ManifeStin ÜYelERi: Sueda, Lidya, Hilal, Mina, Zeynep, Esin!       "

# NOT: Bu işlemlerin hiçbiri sonucu bir değişkene atamadığım için orijinal string'i etkilemez!
print(manifest.lower())
print(manifest.upper())
print(manifest.title())
print(manifest.capitalize())
print(manifest.replace("Zeynep", "Zoktay"))
print(manifest.split(" ", 5))
print(manifest.lower().count("e")) # Birbiri ardına kullanabilirsiniz, sırayla yapar.
print(manifest.count("e"))
print(len(manifest))
print(len(manifest.strip()))
````
