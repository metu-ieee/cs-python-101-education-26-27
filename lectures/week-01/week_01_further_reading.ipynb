#### Complex Veri Tipi
Lisede kompleks/imajiner sayıları görmüştük (Yani ben gördüm ve o kadar yaşlı değilim!). Bu sayılar normalde "a + bi" olarak ifade ediliyordu. Python'da ise bunu "a + bj" olarak ifade edebiliyoruz!

#### Binary, Octal ve Hexadecimal dönüşümler
Elinizde bir sayı varsa ve bunu bir sebepten ötürü bu biçimlerden birine dönüştürmek istiyorsanız yapması çok basit! Her biri için şu şekilde iki yönlü dönüşümler yapabilirsiniz:
* **Binary:** `bin()` fonksiyonu içine aldığı sayıyı binary yapıp string olarak döndürür. Geri almak için çıktıyı int() içinde `int(çıktı, 2)` şeklinde kullanabilirsiniz.
* **Octal:** `oct()` fonksiyonu içine aldığı sayıyı octal yapıp string olarak döndürür. Geri almak için çıktıyı int() içinde `int(çıktı, 8)` şeklinde kullanabilirsiniz.
* **Hexadecimal:** `hex()` fonksiyonu içine aldığı sayıyı hex yapıp string olarak döndürür. Geri almak için çıktıyı int() içinde `int(çıktı, 16)` şeklinde kullanabilirsiniz.

````python
my_int = 27
my_int_bin = bin(my_int)
my_int_oct = oct(my_int)
my_int_hex = hex(my_int)

# Hadi printleyelim!
print(my_int_bin)
print(my_int_oct)
print(my_int_hex)

# Bu variable'ları biribirine input olarak veremiyoruz, yani my_int_hex = hex(my_int_bin) gibi ifadeler hata verir.
# Çünkü artık hepsi bir string, hepsinin başında yer alan 0b, 0o ve 0x bizim birbirine input olarak vermemizi engelliyor, hadi kanıtlayalım:
print(type(my_int_bin))
````

### Part 4: XOR GATE Açıklaması

Ders sırasında XOR GATE yaparken ilginç bir şeye rastlamıştık. `xor = (my_num and not my_num2) or (not my_num and my_num2)` kodunda `my_num` 1'e eşitken cevap `my_num2`'ye bağlı olarak *True* ya da *False* dönerken `my_num` 0 iken cevap 0 ya da 1 olarak dönmekteydi. Bunun sebebi ise `and` ve `or` gate'lerinin diğer dillerde olduğu gibi Python'da da oldukça tembel olmasından kaynaklı. Hadi beraber keşfedelim!

#### And ve Or Nasıl Tembel?

Biz bunu cidden de lazy (tembel) evaluation olarak adlandırıyoruz. Oldukça akıllı bir yöntem. Bu sistem ikisi için şu şekilde çalışıyor:
* `and` gate öncesinde yer alan ifade *False* ise Python sonraki ifadeyi incelemez, direkt ilk ifadeyi döndürür. Örneğin `X and Y` ifadesinde *X* *False*'a denkse *X* döner (Dikkat, *False* dönmez *X* döner!).
* `or` gate öncesinde yer alan ifade *True* ise Python sonraki ifadeyi incelemez, direkt ilk ifadeyi döndürür. Aksi takdirde direkt sonraki ifadeyi döndürür. Örneğin `X or Y` ifadesinde *X* *True*'ya denkse *X*, *False*'a denkse *Y* döner (Dikkat, bool dönmez, *X* ya da *Y* döner!).

**Peki biz bunu normalde neden hiç fark etmiyoruz?**

Çünkü genel kullanımda yukarıda örneklenen X ve Y ifadeleri direkt Ture ya da False'a tekabül eden örnekler oluyor. Biz ise burada 0 ve 1 sayılarını verdik. 
* `my_num` 1 iken xor'da ilk kısımda and'in devamındaki `not my_num2` ifadesi `my_num2`'nin tersini alırken onu bir bool'a a dönüştürüyor. `my_num2` direkt 0 ise dönen *True* or'un devamına bakılmadan dönüyor.
* Yukarıdaki senaryoda `my_num2` 1 ise ilk kısım *False* olup or'un devamına bakıyoruz. `not my_num` *False*'a dönüştüğü ve dolayısıyla and'in devamına bakılmadığı için dönen sonuç *False* oluyor.
* `my_num` 0 iken ise ilk kısımda direkt 0 dönüyor (False değil), or'un devamına bakılıyor. `not my_num` *True*'ya dönüştüğü için `my_num2` ne ise o döner, 0 ya da 1.

````python
# Hatta bu durumu biraz daha örnekleyelim!

# Bu ifadenin daha yazıldığını daha ileriki derslerde öğreneceğiz, ancak şunu bilmeniz yeterli, artık xor iki adet veri alıp bir sonuç dönebiliyor!
def xor(x,y):
    return (x and not y) or (not x and y)

print("Normal kullanım:", xor(True, False)) # Normalde XOR True/False ya da benzer sorgular almalı.
print("Normal kullanım 2:", xor(5< 10, 10 <= 20)) # Normalde XOR ve benzeri gate'ler bu tür bool döndüren sorgular ile kullanılmalı.

# Biz ise XOR'u derste böyle kullandık
print("Ders örneği:", xor(1,0)) 
print("Ders örneği 2:", xor(0,0))

# Biraz farklı örneklerle yukarıda bahsettiğimiz şeyi kanıtlayalım:
print("Tuhaf örnek:", xor("",3))
print("Tuhaf örnek 2:", xor(2,False))
print("Tuhaf örnek 3:", xor(0,"")) # Boş string döndü
````

### Part 5: Float'lar ve Sorunları

Float'lar bizim için her ne kadar küsürlü sayıları tutsa da çok bir problem mevcut. Yukarıda da söylediğimiz gibi, bilgisayarlar 2'lik sistemde sayı tuttuğu için noktadan sonraki kısmı normal bir şekilde saklayamıyoruz. Bunun anlamak için önce integer'ları anlamamız lazım!

#### Integer'lar Nasıl Tutulur?

Günümüzde integer'lar 2'nin tamlayanı yöntemi ile tutulur. Bu yöntemde amaç bir sayıyı negatif hali ile toplayınca sıfır elde etmektir. Bu sistemde en önemli bit (yani en soldaki) 1 olduğu zaman sayılar negatif'e döner. Örneğin, 4 bit'lik bir sistemde sayılar 0000 yani 0 ile başlar. 0001 = 1, 0010 = 2 diye gider. 0111 = 7 sayısına gelindiğinde bir sonraki sayı 1000 = 8 değil -8 olur. Sayılar 1001 = -7, 1010 = -6 diye gider. Bu sistem şu avantajı sağlar:
* Bir sayı negatifi ile toplandığında 0 verir, örneğin 0110 = 6 ile 1010 = -6 toplandığında sonuç 10000 olur, en soldaki 1 5. bit olduğu için 4 bitlik sistemde kayar (overflow) ve sonuç 0000 = 0 olur.
* Bir sayıyı pozitiften negatife ya da negatiften pozitife döndürme işlemi çok daha basit olur, sayının tüm bitleri değiştirip (bit flip) 1 ekleyerek elde ederiz. Örneğin, 0101 = 5'i ter çevirip bir eklersek, 1010 + 1 = 1011 = -5 olur. Aynı şekilde -5'i ters çevirip bir eklersek 0100 + 1 = 0101 = 5 olur.
* Diğer yöntemlere kıyasla tek bir sıfır bulunur (evet bazı sayı tutma yöntemlerinde +0 ve -0 bulunur!) ve pozitif sayılar oldukça anlaşılırdır.

Integer'ları anlamak negatif sayılar ilk başta biraz tuhaf gelse de oldukça kolay. Hadi beraber float'ı inceleyelim.

#### Float'lar Nasıl Tutulur?

Modern sistemler float'ı IEEE 754 standardı (evet burada da IEEE farkı) ile tutar. Bu standart uyarında tek hassasiyet (single precision) (4 byte = 32 bit) ve çift hassasiyet (double precision) (8 byte = 64 bit) olmak üzere iki farklı sistem mevcuttur. Biz şimdi single precision öğreneceğiz, double precision aynı mantığı daha çok bit ile uygulamaktadır. 

Single precision'da yer alan 32 bit 3 parçaya ayrılır:
1. İlk bit işaret bitidir. Sayı pozitif ise 0, negatif ise 1 olarak belirlenir.
2. Sonraki 8 bit exponent (üs) için ayrılmış bitlerdir. Bunlar sayının 2'nin hangi üssü ile çarpılacağını belirtir. 
3. Geriye kalan 23 bitlik kısım *mantissa* olarak adlandırılır. Burada asıl sayı yer alır.

![IEEE_float](https://media.geeksforgeeks.org/wp-content/uploads/Single-Precision-IEEE-754-Floating-Point-Standard.jpg)

Hadi beraber 85.125 sayısını bu sistemde yazmaya çalışalım!
1. Biliyoruz ki 85 binary'de 1010101 olarak yazılıyor, bu cepte.
2. Noktadan sonraki kısmı yazmak için şunu bilmemiz gerekiyor: 10'luk sistemde noktadan sonra 10<sup>-1</sup>, 10<sup>-2</sup> gidiyorsa, 2'lik sistemde de 2<sup>-1</sup>, 2<sup>-2</sup> şeklinde gitmekte. 0.125 de baktığımız zaman 2<sup>-3</sup> yani $1/8$, o zaman 0.125 = 0.001 diyebiliriz
3. Sayıyı birleştirirsek 85.125 = 1010101.001 elde ederiz! Bunu aynı şekilde 1.010101001 x $2^6$ olarak ifade edebiliriz!
4. Sayı pozitif olduğu için sign bit = 0 olarak yazabiliriz!
5. Exponent için ise az önce elde ettiğimiz sayıya bias ekliyoruz (ne olduğuna ve neden olduğuna bakmak için [tıklayabilirisiniz](https://en.wikipedia.org/wiki/Exponent_bias)). Sonuç olarak 127 (bias) + 6 (yukarıda bulduğumuz exponent) toplayıp 133 elde ediyoruz. Yani exponent bitleri 133 = 10000101 olarak yazabiliriz!
6. Mantissa kısmına geldik! Burada IEEE standardı çok mantıklı bir eylemde bulunuyor, her sayının binary yazımında ilk bit doğal olarak bir olacağı için (bunu şöyle de düşünebilirsiniz, 10'luk sistemde sayıları 34 yazıyoruz 0034 yazmıyoruz) ilk basamak atılır ve yukarıda elde ettiğimiz sayının noktadan sonra gelen kısmı Mantissa için yazılır. Yani 1.010101001'in 010101001 kısmı mantissa olarak yazılır! 23 bitin geri kalanı sağa doğru 0'lar ile doldurulur.

Elde ettiğimiz sayı:
0 10000101 01010100100000000000000
yani 85.125!

**Bir dakika! Biz bunun bir sorun olduğundan bahsetmiştik, peki neden?**

Evet, maalesef float'ları daha efektif bir şekilde bilgisayarda tutamıyoruz ve bu yöntemin de bir sorunu mevcut. 2'nin eksili üsleri, yani $1/2^k$'ların toplamı şeklinde ifade edilemeyen her küsürat binary sistemde devirli sayılara (sürekli devam eden döngülere) sebep olmakta. Float da sınırlı olduğu için, bir basamaktan sonra mevcut sayı değişikliğe uğramakta. Modern programalama dilleri ve Python buna bazı çözümler bulmuş olsa da bu sorun hala her dilde mevcut. Python'ın bu soruna olan çözümlerini ilerleyen derslerimizde göreceğiz.

````python
# Hadi bu sorunu örnekleyelim!
print(0.1 + 0.2) # en klasik ve aynı zamanda favorim olan örnek
# Devirli binary'lerden kaynaklı olarak son basamak bir anda 4 oluyor 0 olması gerekirken!
````
